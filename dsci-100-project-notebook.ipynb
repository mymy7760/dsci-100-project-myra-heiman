```r

# Hisogram to show distribution of numeric value
basketball_tidy_hist <- basketball_tidy |>
ggplot(aes(x = reg_wins_pct)) +
  geom_histogram() +
  labs(
    title = "Distribution of Regular Season Win Percentage",
    x = "Regular Season Win Percentage",
    y = "Number of Teams" )
basketball_tidy_hist
```

```r
# Bar plot for categorial variables

basketball_tidy_bar <- basketball_tidy |>
  count(elite_eight) |>
  mutate(proportion = n / sum(n)) |>
  ggplot(aes(x = elite_eight, y = proportion)) +
  geom_col() +
  labs(
    title = "Proportion of Teams Reaching the Elite Eight",
    x = "Reached the Elite Eight",
    y = "Proportion of Teams")
basketball_tidy_bar
```

This scatterplot shows the relationship between regular-season win percentage and tournament seed. Teams with higher win percentages generally receive better tournament seeds, shown by lower seed numbers. However, there is still variation between teams, suggesting that regular-season win percentage is not the only factor used to determine tournament seeding.

```r
basketball_tidy_point <- basketball_tidy |>
ggplot(aes(x = reg_wins_pct, y = seed)) +
  geom_point() +
  labs(
    title = "Regular Season Win Percentage and Tournament Seed",
    x = "Regular Season Win Percentage",
    y = "Tournament Seed (Lower Number = Higher Seed)")
basketball_tidy_point
```

This scatterplot shows regular season win percentage and tournament seed, while using red and blue to distinguish teams that reached the Elite Eight from those that did not. It helps identify whether teams with higher win percentages and better tournament seeds were more likely to advance.

```r
basketball_tidy_eight <- basketball_tidy |>
ggplot(aes(x = reg_wins_pct, y = seed, colour = elite_eight)) +
  geom_point() +
  labs(
    title = "Regular Season Win Percentage and Elite Eight Performance",
    x = "Regular Season Win Percentage",
    y = "Tournament Seed",
    colour = "Reached Elite Eight"
  )
basketball_tidy_eight
```
